```python
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.graphics.tsaplots import plot_acf
```

```python
veri = np.random.randn(1000)
```

```python
plt.plot(np.arange(len(veri)), veri)
plt.title("Beyaz Gürültü Süreci")
```

```python
parcalar = np.split(veri, 10)
ort, var = [], []

for i in parcalar:
    ort.append(np.mean(i))
    var.append(np.var(i))

plt.plot(np.arange(len(ort)), len(ort)*[veri.mean()], label="Ortalama", color="blue")
plt.scatter(x=np.arange(len(ort)), y=ort, color="blue")
plt.plot(np.arange(len(var)), len(var)*[veri.var()], label="Varyans", color="red")
plt.scatter(x=np.arange(len(var)), y=var, color="red")
plt.show()
```
