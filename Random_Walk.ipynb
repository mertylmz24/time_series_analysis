```python
import random
import matplotlib.pyplot as plt
from statsmodels.graphics.tsaplots import plot_acf
import pandas as pd
import numpy as np
from arch.unitroot import ADF

def rw(n):
    x,y = 0,0
    yon = ["İ", "G"]
    konum = [y]
    zaman = np.arange(n+1)
    
    for i in range(1, n+1):
        adim = random.choice(yon)
        if adim == "İ":
            y=y+1

        else:
            y=y-1
        konum.append(y)
        
    return zaman, konum

rw1 = rw(500)
rw2 = rw(500)
rw3 = rw(500)



plt.plot(rw1[0], rw1[1], color="blue", label="rw1")
plt.plot(rw2[0], rw2[1], color="red", label="rw2")
plt.plot(rw3[0], rw3[1], color="green", label="rw3")
plt.legend()
plt.show()




    
    
```
